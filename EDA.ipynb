### House Prediction

### Problem Statement

The Mumbai real estate market is one of the most expensive and complex in India, where property prices are influenced by several factors such as BHK, area, locality, region, construction status, and age of the property. This dataset of over 76,000 listings provides detailed information on these attributes, offering an opportunity to analyze how they affect housing prices. The main objective is to identify key factors driving property costs, compare trends like ready-to-move vs. under-construction homes, and highlight price variations across regions. Such insights can help buyers, sellers, investors, and real estate companies make better and more informed decisions.

### Heatmap of Correlation

```python


# Copy dataframe to avoid changes in original
df_clean = df.copy()

# 1. Convert datetime columns (if any) into numeric (timestamp)
for col in df_clean.select_dtypes(include=['object', 'datetime']):
    try:
        df_clean[col] = pd.to_datetime(df_clean[col], errors='coerce')
        if pd.api.types.is_datetime64_any_dtype(df_clean[col]):
            df_clean[col] = df_clean[col].astype(int) // 10**9  # UNIX timestamp
    except:
        df_clean.drop(columns=[col], inplace=True)  # agar convert na ho to drop

# 2. Select only numeric columns
numeric_df = df_clean.select_dtypes(include=['number'])

# 3. Correlation matrix
corr = numeric_df.corr()

# 4. Plot heatmap
plt.figure(figsize=(10,8))
sns.heatmap(corr, vmax=1, square=False, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap", fontsize=14)
plt.show()
```
